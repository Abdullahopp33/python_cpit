```python
import numpy as np
import matplotlib.pyplot as plt
import pandas as pd

df = pd.read_csv("Iris.csv")

# Q 1: Plot a histogram of Sepal Length
plt.hist(df['SepalLengthCm'], bins=10)
plt.title("Histogram of Sepal Length")
plt.xlabel("Sepal Length (cm)")
plt.ylabel("Frequency")
plt.show()

# Q 2: Plot a boxplot of all the features
df.boxplot(column=['SepalLengthCm', 'SepalWidthCm', 'PetalLengthCm', 'PetalWidthCm'])
plt.title("Boxplot of Iris Features")
plt.show()

# Q 3: Display the first 10 rows of the dataset
print(df.head(10))




```
