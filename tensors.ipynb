```python
import tensorflow as tf
from tensorflow.keras import layers
import numpy as np

a = np.array([1,2,3,4,5])
b = np.array([2,4,6,8,10])

model = tf.keras.Sequential([
    layers.Dense (units =1 , input_shape = [1])
])

model.compile(optimizer ='sgd', loss='mean_squared_error') #SGD = Stochastic Gradient Desent
model.fit(a,b, epochs=5)
print(model.predict(np.array([6.0])))
```

```python
import tensorflow as tf
from tensorflow.keras import layers
import numpy as np

# Create a simple dataset
a = np.array([1,2,3,4,5])
b = np.array([2,4,6,8,10])

model = tf.keras.Sequential([
    layers.Dense(1, input_shape=(1,))
])

model.compile(optimizer='sgd', loss="mse")
model.fit(a, b, epochs=10,)

print(model.predict(np.array([[6.0]])))
```

```python
import tensorflow as tf
from tensorflow.keras import layers
import numpy as np

# Create a simple model
a = np.array([1,2,3,4,5])
b = np.array([2,4,6,8,10])

model = tf.keras.Sequential([
    layers.Dense(20, activation='relu', input_shape=[1]),
    layers.Dense(1, activation='relu', kernel_initializer='uniform'),
    layers.Dense(1, activation='relu')
])

model.compile(optimizer='sgd', loss="mse")
model.fit(a, b, epochs=5,)

print(model.predict(np.array([6.0])))

```

```python
import tensorflow as tf
from tensorflow.keras import layers
import numpy as np

# Create a simple model

a = np.array([
    [1,2],
    [3,4],
    [1,4],
    [5,6],
    [1,2],
    [4,2]

], dtype=np.float32 )

b = np.array([0,1,0,1,0,2], dtype=np.float32)

model = tf.keras.Sequential([
    layers.Dense(8, activation='relu', input_shape=(2,)),
  #  layers.Dense(1, activation='relu', kernel_initializer='uniform'),
    layers.Dense(3, activation='softmax')
])

model.compile(
    optimizer="adam",
    loss="sparse_categorical_crossentropy", metrics=['accuracy'] ) 
model.fit(a,b, epochs=5 )

test = np.array([[2.0,3.0],[1.0,2.0],[4.0,1.0]])
predictions = model.predict(test)

print(predictions)
print( np.argmax(predictions,axis=1))

# print(model.predict(np.array([[6.0,7.0]])))
```

```python
import tensorflow as tf
from tensorflow.keras import layers
import numpy as np
import matplotlib.pyplot as plt

(x_train, y_train), (x_test, y_test) = tf.keras.datasets.mnist.load_data()

x_train = x_train / 255.0
x_test = x_test / 255.

model = tf.keras.Sequential([
    layers.Flatten(input_shape=(28,28)),
    layers.Dense(128, activation='relu'),
    layers.Dense(10, activation='softmax')
])

model.compile(
    optimizer="adam",
    loss="sparse_categorical_crossentropy",
    metrics=['accuracy']
) 
model.fit(x_train,y_train, epochs=5 )

history = model.fit(x_train,y_train, epochs=5, validation_data=(x_test,y_test))

plt.plot(history.history['accuracy'], label='accuracy')
# plt.plot(history.history['val_accuracy'], label='val_accuracy')
plt.xlabel('Epoch')
plt.ylabel('Accuracy')
test_loss, test_acc = model.evaluate(x_test, y_test)
print(test_acc * 100 )


```
