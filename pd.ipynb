```python

months = ["Jan", "Feb", "Mar", "Apr", "May", "Jun", "Jul", "Aug", "Sep", "Oct",None, "Dec"]
sales = np.array([1000, 1500, 1200, 1800, 2000,None, 2500, 2200, 2800, 3000, 3500, 4000])
df_sales = pd.DataFrame({"Month": months, "Sales": sales})
df_sales.dropna(inplace=True)
print(df_sales)
```

```python
avg1 = df_sales["Sales"].mean()
print("Average Sales:", avg1)

max1 = df_sales["Sales"].max()
print("Maximum Sales:", max1)

min1 = df_sales["Sales"].min()
print("Minimum Sales:", min1)

best_month = df_sales[df_sales["Sales"] == max1]
print("Best Month for sales:", best_month)

best_month = df_sales[df_sales["Sales"] == min1]
print("Best Month for sales:", best_month)

best_month =df_sales.loc[df_sales["Sales"].idxmax(), "Month"]
print("Best Month for sales:", best_month)

best_month =df_sales.loc[df_sales["Sales"].idxmin(), "Month"]
print("Best Month for sales:", best_month)
```

```python
# Employee Salary Analysis 5 employee 5 salaries find max salary,min salary and average salary with employee name.

employees_names = ["Ali", "Saad", "Usman", "Daud", "Ahmed"]
salaries = np.array([50000, 60000, 55000, 70000, 65000])

df_employees = pd.DataFrame({"Employee": employees_names, "Salary": salaries})
print(df_employees)
```

```python
max_salary = df_employees["Salary"].max()
print("Maximum Salary:", max_salary)

min_salary = df_employees["Salary"].min()
print("Minimum Salary:", min_salary)

avg_salary = df_employees["Salary"].mean()
print("Average Salary:", avg_salary)
```

```python
max_salary = df_employees["Salary"].max()
max_emp_idx = df_employees["Salary"].idxmax()
max_emp_name = df_employees.loc[max_emp_idx, "Employee"]
print(f"Employee with Max Salary: {max_emp_name}, Salary: {max_salary}")

min_salary = df_employees["Salary"].min()
min_emp_idx = df_employees["Salary"].idxmin()
min_emp_name = df_employees.loc[min_emp_idx, "Employee"]
print(f"Employee with Min Salary: {min_emp_name}, Salary: {min_salary}")
```
