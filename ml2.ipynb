```python
from sklearn.linear_model import LinearRegression
import numpy as np

# size of the house in square feet # independent in 2D array
house = np.array([100,150,200,250,300])
# price of the house in dollars # dependent in 1D array
price = np.array([200000,300000,400000,500000,600000])

model = LinearRegression()
model.fit(house.reshape(-1, 1), price) # reshape(-1, 1) is used to convert the 1D array into a 2D array with one column

# predict the price of a house with size 350 square feet
predicted_price = model.predict([[350]]) # predict() method expects a 2D array as input, so we pass [[350]] instead of [350]
print(predicted_price)
```
