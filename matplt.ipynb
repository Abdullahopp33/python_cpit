```python
a = np.array([1, 2, 3, 4, 5])
b = np.array([2,3,4,2,6])
c = np.array([1,2,3,4,5])

plt.title("Linear Regression")
plt.xlabel("X-axis")
plt.ylabel("Y-axis")

plt.plot(b, a, color="red" , marker="o", mec="blue" ,mfc="yellow", ms=10, ls="-", lw=3) # "o" is used to plot the points as circles and o- is used to plot the points as circles and connect them with lines and D is used to plot the points as diamonds and D- is used to plot the points as diamonds and connect them with lines and square and * too
plt.plot(a, c, color="blue")
plt.show()

# mec = marker edge color means the color of the edge of the marker
# mfc = marker face color  means the color of the face of the marker
# ms = marker size means the size of the marker
# ls = line style means the style of the line and it can be -- for dashed line and - for solid line and : for dotted line and -. for dash-dot line
# lw = line width means the width of the line
```

```python
import matplotlib.pyplot as plt
import numpy as np
import pandas as pd

data = pd.read_csv("Housing.csv")

price = data["price"].head(20)
area = data["area"].head(20)

plt.title("Price vs Area")
plt.xlabel("Area")
plt.ylabel("Price")
plt.plot(area, price)
plt.show()

plt.bar(area, price ,width=9)
plt.show()
```

```python
df = pd.read_csv("country_wise_latest.csv")

countries = df["Country"].head()
deaths = df["Deaths"].head()

plt.xticks(rotation=45) # xticks is used to rotate the x-axis labels and it takes one argument which is the angle of rotation
plt.yticks(rotation=45) # yticks is used to rotate the y-axis labels and it takes one argument which is the angle of rotation


plt.bar(countries, deaths, color=["red", "blue", "green", "yellow", "orange"])
plt.show()
```

```python
df = pd.read_csv("country_wise_latest.csv")

countries = df["Country"].head()
deaths = df["Deaths"].head()
recovered = df["Recovered"].head()

plt.xticks(rotation=45) # xticks is used to rotate the x-axis labels and it takes one argument which is the angle of rotation
plt.yticks(rotation=45) # yticks is used to rotate the y-axis labels and it takes one argument which is the angle of rotation


# plt.figure(figsize=(9, 9)) # (Ye output nhi dy ga)figure is used to set the size of the plot and it takes one argument which is a tuple of the width and height of the plot in inches
plt.bar(countries, recovered, color='red', label="Recovered") # label is used to set the label for the legend and it takes one argument which is the label name
plt.bar(countries, deaths, color='blue', label="Deaths") # label is used to set the label for the legend and it takes one argument which is the label name
plt.legend() # legend is used to display the legend and it takes no arguments

plt.savefig("covid.png") # savefig is used to save the plot as an image and it takes one argument which is the name of the image file and it can be in png, jpg, jpeg, svg, pdf, etc. formats
plt.show()
```
